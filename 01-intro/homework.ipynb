```python
mostFreqHorsepower = df['horsepower'].value_counts().keys()[0]
print(mostFreqHorsepower)
```

```python
df.fillna({'horsepower': mostFreqHorsepower}, inplace = True)
df['horsepower'].median()
```

```python
X = df.loc[df['origin'] == 'Asia', ['vehicle_weight','model_year']].iloc[0:7].to_numpy()
print(X)
```

```python
XT = X.T
print(XT)
```

```python
XTX = XT @ X
print(XTX)
```

```python
XTXinv = np.linalg.inv(XTX)
print(XTXinv)
```

```python
y = np.asarray([1100, 1300, 800, 900, 1000, 1100, 1200])
print(y)
```

```python
w = XTXinv @ XT @ y
print(w)
```
